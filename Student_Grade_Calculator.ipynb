# Student Grade Calculator

A beginner Python project that calculates the average of three test scores, assigns a letter grade, and displays pass/fail status.

### Features
- Accepts a student's name and three equally weighted test scores
- Accepts decimal scores between 0 and 100
- Repeats the current prompt after out-of-range or non-numeric input
- Displays the average, letter grade, and pass/fail status

### Grading rules used in this project
- A: average >= 80
- B: 70 <= average < 80
- C: 60 <= average < 70
- D: 50 <= average < 60
- F: average < 50
- Pass: average >= 50; Fail: average < 50

The grade and status use the unrounded average. The displayed average is rounded to two decimal places.

### Run
Run the code cell below with Shift + Enter and answer its prompts. See README.md for setup and TEST_CASES.md for the validation results.

### Built with
Python and Jupyter Notebook. The calculator uses only Python's built-in functions.


```python
print("Student Grade Calculator")

name = input("Enter student name: ")

def get_score(number):
    while True:
        try:
            score = float(input(f"Enter score {number}: "))

            if 0 <= score <= 100:
                return score
            else:
                print("Please enter a score between 0 and 100.")

        except ValueError:
            print("Please enter a valid number.")

score1 = get_score(1)
score2 = get_score(2)
score3 = get_score(3)

average = (score1 + score2 + score3) / 3

if average >= 80:
    grade = "A"
elif average >= 70:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 50:
    grade = "D"
else:
    grade = "F"

print("\n----- Result -----")
print("Student:", name)
print("Average:", round(average, 2))
print("Grade:", grade)

if average >= 50:
    status = "Pass"
else:
    status = "Fail"
print("Status:", status)
```
